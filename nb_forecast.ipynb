```python
import numpy as np
import pyccl as ccl
from bao_forecast import BAOBin, combined_bao_error

# ---------------------------------------------------------
# Cosmology
# ---------------------------------------------------------

cosmo = ccl.Cosmology(
    Omega_c=0.25,
    Omega_b=0.05,
    h=0.7,
    sigma8=0.8,
    n_s=0.96,
)

# ---------------------------------------------------------
# Data setup
# ------=----------------------------------------------------

z_min = 0.6
z_max = 1.2
dz = 0.1
nz = [2854542, 3266097, 3898672, 3404744, 1752169, 761332]
sz = [0.0232, 0.0254, 0.0292, 0.0358, 0.0403, 0.0415]
area = 4273.42

# z_min = 0.6
# z_max = 1.1
# dz = 0.1
# nz = [1478178, 1632805,  1727646, 1315604,  877760]
# sz = [0.021, 0.025, 0.029, 0.030, 0.040]
# area = 4108.47

# area = 12300
# # area = 14000
# z_min = 0.2
# z_max = 1.2
# dz = 0.2
# nz = np.array([3.11, 4.29, 4.25, 3.59, 2.76]) * (area * 60**2)
# sz = [0.03] * len(nz)

# ---------------------------------------------------------
# Survey bins
# ---------------------------------------------------------

bins = [BAOBin(z_min + dz*i, z_min + dz*(i+1), nz[i], sz[i])
    for i in range(len(nz))
]

# ---------------------------------------------------------
# Forecast
# ---------------------------------------------------------

forecast = combined_bao_error(
    bins=bins,
    area_deg2=area,
    cosmo=cosmo,
)

print(f"{forecast * 100:.3f}%")
```
