```python
# LSST Y10

area = 14300
# area = 18000
zmin = 0.2
zmax = 1.2
dz = 0.1
nz = np.array([2.89, 4.09, 4.94, 5.45, 5.65, 5.61, 5.4, 5.07, 4.67, 4.23]) * (area * 60**2)
sz = [0.03] * len(nz)

forecast = BAOerr4DES.Y1tot(zmin, zmax, dz, nz, sz, area)
print(forecast * 100 * 1.5)

# There is barely any improvement because the forecast is close to the
# cosmic-variance limit, so increasing the galaxy density from Y1 to Y10 barely
# helps. The forecasted error has two competing contributions: cosmic variance and
# shotnoise. The variance of a Fourier mode is proportional to (P(k)+1/n)**2, and our
# code uses V_eff = (nP/(1+nP))**2*V. If nP<<1, then V_eff ~ (nP)**2*V and sigma_BAO
# is proportional to 1/nP, which means increasing the galaxy density strongly improves 
# the constraints. However, if nP>>1, then V_eff --> V, and sigma_BAO is proportional to V**(-1/2)
```
